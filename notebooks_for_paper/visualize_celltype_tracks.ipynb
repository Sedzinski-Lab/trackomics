```python
import os
import pandas as pd
import numpy as np
```

```python
from tifffile import imread
import napari

# Load nuclei timelapse file
image = imread('D:/Mari_Sixth_Dataset_Analysis/nuclei_timelapses/timelapse_sixth_dataset.tif')
viewer = napari.view_image(image, name='nuclei', colormap='gray')

napari.run()
```

```python
full_df = pd.read_csv("D:/Mari_Sixth_Dataset_Analysis/nuclei_membrane_tracking/full_manual_dataset_cropped.csv")
```

```python
full_df2 = full_df[full_df['cell_type'] != 'unknown']
```

```python
for cell_type in full_df2['cell_type'].unique():
    subset_df = full_df2[full_df2['cell_type'] == cell_type]
    subset_df['t_frame'] = subset_df['Spot frame'].astype(int) + 1
    print(f"Cell type: {cell_type}, Number of tracks: {subset_df['Spot track ID relabelled'].nunique()}")
    # Load and add tracks
    df = subset_df.rename(columns={
        'Track ID_y': 'track_id', 't_frame': 'frame',
        'Z_orig': 'z', 'Y_orig': 'y', 'X_orig': 'x'
    }).sort_values(by=['track_id', 'frame'])

    tracks_data = df[['track_id', 'frame', 'z', 'y', 'x']].to_numpy()
    # Make a constant-valued property array (e.g., all ones)
    n_rows = tracks_data.shape[0]
    ones = np.ones(n_rows)  # or any value or pattern
    ones[0] = 0.0000001

    properties = {
        'ones': ones
    }

    viewer.add_tracks(tracks_data, properties = properties, name=f'{cell_type} tracks')
```

Select colorization and track layers visible in Napari, then run the frame saving code below

```python
from PIL import Image
for t in range(0, 360):
    # Update viewer to display the current subarray
    viewer.dims.current_step = (t, 8, 1071, 1071)
    img = viewer.screenshot(scale=1, canvas_only=True, flash=False)
    # Convert the NumPy array to a Pillow Image
    image = Image.fromarray(img, mode="RGBA")
    
    # Save the image as a PNG
    save_path = f'D:/Mari_Sixth_Dataset_Analysis/animation_tracks/t{t}.png'
    image.save(save_path, format="PNG")
```
